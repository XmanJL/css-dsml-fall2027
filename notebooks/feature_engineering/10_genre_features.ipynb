### Columns to take from processed sets: `id`, `production_companies`, `production_countries`, `spoken_languages`, `director`, `top_cast`, `keywords`, `title`, `original_title`,  `original_title`, `original_language`, `overview`, `status`
### Additonal columns: `belongs_to_collection`

### New columns: `has_collection`, `company_count`, `country_count`, `spoken_language_count` | `cast_size`, `crew_size`, `has_director` | `keyword_count`, `title_word_count`, `overview_word_count`, `title_differs_from_original`

```python
print(metadata_df['id'].duplicated().sum())
metadata_df.drop_duplicates(subset='id', inplace=True)
print(metadata_df['id'].duplicated().sum())

print(metadata_df['id'].isna().sum())
metadata_df.dropna(inplace=True)
print(metadata_df['id'].isna().sum())

print(metadata_df['id'].dtype)
metadata_df['id'] = metadata_df['id'].astype(int)
print(metadata_df['id'].dtype)

display(metadata_df['id'].info())
```

```python
data_frames = [production_df, credits_df, keywords_df, text_df]
combined_df = reduce(lambda  left,right: pd.merge(left,right,on=['id'], how='outer'), data_frames)
display(combined_df.head())
print()
display(combined_df.info())
```

```python
combined_df.drop(columns=['original_title_x', 'release_date'], inplace=True)
combined_df.rename(columns={'original_title_y': 'original_title'}, inplace=True)
display(combined_df.info())
```

```python
final_df = combined_df.copy()
```

```python
final_df['has_collection'] = combined_df['belongs_to_collection'].notna()
display(final_df[['belongs_to_collection', 'has_collection']])
```

```python
final_df['company_count'] = combined_df['production_companies'].str.split("|").str.len()
final_df['company_count'] = pd.to_numeric(final_df['company_count'], errors='coerce').fillna(0).astype(int)
display(final_df[['company_count', 'production_companies']])
```

```python
final_df['country_count'] = combined_df['production_countries'].str.split("|").str.len()
final_df['country_count'] = pd.to_numeric(final_df['country_count'], errors='coerce').fillna(0).astype(int)
display(final_df[['country_count', 'production_countries']])
```

```python
final_df['spoken_languages_count'] = combined_df['spoken_languages'].str.split("|").str.len()
final_df['spoken_languages_count'] = pd.to_numeric(final_df['spoken_languages_count'], errors='coerce').fillna(0).astype(int)
display(final_df[['spoken_languages_count', 'spoken_languages']])
```

```python
final_df['has_director'] = combined_df['director'].notna()
display(final_df[['has_director', 'director']])
```

```python
final_df['keyword_count'] = combined_df['keyword_names_clean'].dropna().apply(lambda x: len(ast.literal_eval(x)))
final_df['keyword_count'] = pd.to_numeric(final_df['keyword_count'], errors='coerce').fillna(0).astype(int)
display(final_df[['keyword_count', 'keyword_names_clean']])
```

```python
final_df['title_word_count'] = combined_df['title'].str.split().str.len()
final_df['title_word_count'] = pd.to_numeric(final_df['title_word_count'], errors='coerce').fillna(0).astype(int)
display(final_df[['title_word_count', 'title']])
```

```python
final_df['overview_word_count'] = combined_df['overview'].str.split().str.len()
final_df['overview_word_count'] = pd.to_numeric(final_df['overview_word_count'], errors='coerce').fillna(0).astype(int)
display(final_df[['overview_word_count', 'overview']])
```

```python
final_df['title_differs_from_original'] = np.where(combined_df['title'] != combined_df['original_title'], True, False)
display(final_df[['title_differs_from_original', 'title', 'original_title']])
```

```python
all_genres = combined_df['genres'].dropna().str.split("|").explode().unique()
print(all_genres)
```

```python
for genre in all_genres:
    final_df[genre] = combined_df['genres'].str.contains(genre).astype(int)

columns_to_print = ['genres'] + list(all_genres)
display(final_df[columns_to_print])
```

```python
final_df.drop(columns=['genres'], inplace=True)
final_df.rename(columns={'keyword_names_clean': 'keywords', 'top_five_cast': 'top_cast'}, inplace=True)
```

```python
#G

# Load the metadata again so we preserve rows with missing feature values
metadata_features_df = pd.read_csv(
    "../../data/processed/movies_metadata.csv"
)

# Prepare movie IDs for merging
metadata_features_df["id"] = pd.to_numeric(
    metadata_features_df["id"], errors="coerce"
)
metadata_features_df = metadata_features_df.dropna(subset=["id"])
metadata_features_df["id"] = metadata_features_df["id"].astype(int)
metadata_features_df = metadata_features_df.drop_duplicates(
    subset=["id"]
)

# Convert release_date to a date format
metadata_features_df["release_date"] = pd.to_datetime(
    metadata_features_df["release_date"], errors="coerce"
)

# Create the date features
metadata_features_df["release_year"] = (
    metadata_features_df["release_date"].dt.year
)
metadata_features_df["release_month"] = (
    metadata_features_df["release_date"].dt.month
)
metadata_features_df["release_decade"] = (
    metadata_features_df["release_year"] // 10 * 10
).astype("Int64")

# Convert numeric columns to numbers
for column in ["budget", "revenue", "runtime", "popularity"]:
    metadata_features_df[column] = pd.to_numeric(
        metadata_features_df[column], errors="coerce"
    )

# Create log features; treat negative values as invalid
metadata_features_df["log_budget"] = np.log1p(
    metadata_features_df["budget"].where(
        metadata_features_df["budget"] >= 0
    )
)
metadata_features_df["log_revenue"] = np.log1p(
    metadata_features_df["revenue"].where(
        metadata_features_df["revenue"] >= 0
    )
)
metadata_features_df["log_popularity"] = np.log1p(
    metadata_features_df["popularity"].where(
        metadata_features_df["popularity"] >= 0
    )
)

# Keep only the ID and the six features we need
feature_columns = [
    "id",
    "release_year",
    "release_month",
    "release_decade",
    "log_budget",
    "log_revenue",
    "runtime",
    "log_popularity",
]

metadata_features_df = metadata_features_df[feature_columns]

# Merge the features into the existing final dataset
final_df["id"] = pd.to_numeric(final_df["id"], errors="raise").astype(int)

final_df = final_df.merge(
    metadata_features_df,
    on="id",
    how="left",
    validate="one_to_one"
)

# Check the new columns and confirm row count
print("Final dataset shape:", final_df.shape)
print("Duplicate IDs:", final_df["id"].duplicated().sum())

display(final_df[feature_columns].head())
```
