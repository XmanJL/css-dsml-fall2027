```python
import ast
import pandas as pd

df = pd.read_csv("../../data/raw/keywords.csv")
#df Parsed
dfp = pd.read_json("../../data/processed/keywords_parsed.json", orient="records")
```

```python
#Checking a sample of the keywords column to see what the data looks like
sample = df["keywords"].iloc[0]
print(sample)
print(type(sample))
```

```python
#Testing that sample to ensure parsing works as expected
parsed_sample = ast.literal_eval(sample)
print(parsed_sample)
print(type(parsed_sample))
```

```python
#Parsing entire keywords column
df["keywords_parsed"] = df["keywords"].apply(ast.literal_eval)
```

```python
#Checking to ensure that the parsing worked as expected
df["keywords_parsed"].head()
df["keywords_parsed"].apply(type).value_counts()
```

```python

df.to_json("../../data/processed/keywords_parsed.json", orient="records")
```
